```python
import pandas as pd
```

```python
data = {
    '松田の労働時間': [160, 150],
    '浅木の労働時間': [161, 175]
}

df = pd.DataFrame(data)
df
```

```python
df = pd.read_csv('KvsT.csv')

df.head(3)
```

```python
df['身長']
```

```python
xcol = ['身長', '体重', '年代']
x = df[xcol]
x
```

```python
t = df['派閥']
t
```

```python
from sklearn import tree
```

```python
# モデルの準備
model = tree.DecisionTreeClassifier(random_state = 0)

# 学習の実行（x、tは事前に定義済みの特徴量と正解ラベル）
model.fit(x, t)
```

```python
taro = [[170, 70, 20]]
taro_df = pd.DataFrame(taro,
                       columns=x.columns)

model.predict(taro_df)
```

```python
matsuda = [172, 65, 20]
asagi = [158, 48, 20]
new_data=pd.DataFrame([matsuda,asagi]
                      ,columns=x.columns)

model.predict(new_data)
```

```python
model.score(x, t)
```

```python
import pickle

with open('KinokoTakenoko.pkl', 'wb') as f:
  pickle.dump(model, f)
```
